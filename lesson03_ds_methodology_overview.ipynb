# Lesson 3 – Data Science Methodology Overview
Based on Unit 2 (pp. 17–24).

Objectives:
- Understand the overall DS methodology.
- Map a simple problem to the 10 steps.


## 10-Step Data Science Methodology (high-level)
1. Business Understanding
2. Analytic Approach
3. Data Requirements
4. Data Collection
5. Data Understanding
6. Data Preparation
7. Modelling
8. Evaluation
9. Deployment
10. Feedback

### Exercise
- Pick a simple problem (e.g., predicting marks from study hours)
- Write one line for each step describing what you would do.
