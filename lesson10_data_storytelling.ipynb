# Lesson 10 – Data Storytelling
Based on Unit 8 (pp. 146–168).

Objectives:
- Turn a small dataset into a simple data story.
- Practice choosing visuals and a narrative.


```
import pandas as pd

# Tiny dataset for storytelling
df = pd.DataFrame({
    'Month': ['Jan', 'Feb', 'Mar', 'Apr'],
    'Views': [1000, 1800, 2600, 2500]
})
df
```

### Story draft
- In January, the channel starts at 1,000 views.
- Views grow sharply till March, then flatten slightly.
- This suggests the new content series launched in Feb worked well.
