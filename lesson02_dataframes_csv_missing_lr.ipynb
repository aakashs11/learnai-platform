```
import pandas as pd

# Example DataFrame
df = pd.DataFrame({
    'Name': ['Aman', 'Beena', 'Chirag'],
    'Maths': [90, 85, None]
})
df
```

```
# Handling missing values
df_filled = df.fillna(df['Maths'].mean())
df_filled
```

```
from sklearn.linear_model import LinearRegression
import numpy as np

# Tiny Linear Regression example
X = np.array([[1], [2], [3], [4]])  # hours studied
y = np.array([50, 60, 70, 80])      # marks
model = LinearRegression()
model.fit(X, y)
pred = model.predict([[5]])
pred
```
