Objectives:
- Understand neurons, weights, activation.
- Manually compute a tiny perceptron.


```
# Simple perceptron-style computation
import numpy as np

x = np.array([1.0, 0.5])   # inputs
w = np.array([0.8, -0.4])  # weights
b = 0.1                    # bias

z = np.dot(w, x) + b
def relu(t):
    return max(0.0, t)

z, relu(z)
```
